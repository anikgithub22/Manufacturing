```python
from fastapi import FastAPI
from fastapi.responses import HTMLResponse
from fastapi.staticfiles import StaticFiles
from fastapi.middleware.cors import CORSMiddleware
import uvicorn
import numpy as np

app = FastAPI()

# Allow CORS for frontend interaction
app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)

# Simulate the gas lift system
@app.post("/simulate/")
async def simulate(gas_injection_rate: float):
    pressure_drop = 0.05 * 3000 / 100 + gas_injection_rate * 10 / (np.pi * (0.1 / 2) ** 2)
    production_rate = 500 * (1 + 0.85 * (gas_injection_rate / 2)) * (1 - pressure_drop / 3500)
    production_rate = max(production_rate, 0)

    return {
        "GasInjectionRate": gas_injection_rate,
        "PressureDrop": pressure_drop,
        "ProductionRate": production_rate,
    }

# Serve the web interface
@app.get("/", response_class=HTMLResponse)
async def serve_frontend():
    html_content = """
    <!DOCTYPE html>
    <html>
    <head>
        <title>Gas Lift Well Simulator</title>
        <script>
            async function simulate() {
                const gasInjectionRate = document.getElementById("gasInjectionRate").value;
                const response = await fetch("/simulate/", {
                    method: "POST",
                    headers: { "Content-Type": "application/json" },
                    body: JSON.stringify({ gas_injection_rate: parseFloat(gasInjectionRate) }),
                });
                const result = await response.json();
                document.getElementById("results").innerHTML = `
                    <p><b>Gas Injection Rate:</b> ${result.GasInjectionRate.toFixed(2)}</p>
                    <p><b>Pressure Drop:</b> ${result.PressureDrop.toFixed(2)}</p>
                    <p><b>Production Rate:</b> ${result.ProductionRate.toFixed(2)}</p>
                `;
            }
        </script>
    </head>
    <body>
        <h1>Gas Lift Well Simulator</h1>
        <div>
            <label for="gasInjectionRate">Gas Injection Rate (0-2):</label>
            <input type="range" id="gasInjectionRate" min="0" max="2" step="0.1" value="1" oninput="document.getElementById('rateValue').innerText = this.value;">
            <span id="rateValue">1</span>
        </div>
        <button onclick="simulate()">Simulate</button>
        <div id="results" style="margin-top: 20px;">
            <p>Simulation results will appear here.</p>
        </div>
    </body>
    </html>
    """
    return HTMLResponse(content=html_content)

# Run the server
if __name__ == "__main__":
    uvicorn.run(app, host="0.0.0.0", port=8000)
```
